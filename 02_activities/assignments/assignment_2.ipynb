```python
# Your answer here

'''
The solution uses depth-first search to traverse the binary tree from the root to all leaf nodes.
This method efficiently finds and returns all root-to-leaf paths in the tree because it explores 
each path only once using a depth-first approach, ensures that every possible path is captured. 
It moves deeper into the tree before backtracking to explore other paths.
When a leaf node is encountered, the path is saved in a list of all paths.
Backtracking ensures that each path is explored fully before returning, making the solution both correct and optimal.

'''
```
